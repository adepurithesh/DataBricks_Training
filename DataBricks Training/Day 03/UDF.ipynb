##User Defined Functions

```python
%sql
create function acc_dev.rithesh.full_name(first_name string,Last_name string)
returns string
return concat(first_name,' ',Last_name)
```

```python
%sql
select acc_dev.rithesh.full_name('rithesh','adepu')
```

```python
%sql
create or replace function acc_dev.rithesh.Addition(a int,b int)
returns int
return a+b;
select acc_dev.rithesh.Addition(10,20)
```
